## Ա մաս: `return`

Երեկվա ֆունկցիաները **տպում** էին։ Դա սահմանափակող է։

```python
def print_average(grades):
    total = 0
    for grade in grades:
        total = total + grade
    print(f"Միջինը՝ {total / len(grades):.1f}")

print_average([9, 10, 8])
```

Սա տպեց միջինը։ Բայց հիմա փորձի՛ր պատասխանել հարցին՝ «արդյո՞ք Անիի միջինը 8-ից բարձր է»։

Չես կարող։ **Թիվը գնաց էկրան, ոչ թե քեզ։**

Լուծումը՝ `return`։

```python
def average_of(grades):
    total = 0
    for grade in grades:
        total = total + grade
    return round(total / len(grades), 1)
```

Այս բջիջը ոչինչ չտպեց, և կանչելիս էլ ոչինչ չի տպի — այն **վերադարձնում** է արժեքը։

```python
average_of([9, 10, 8])
```

Տետրը ցույց տվեց արժեքը, որովհետև այն բջիջի վերջին տողն էր (5-րդ օրվա հնարքը)։

Բայց իրական տարբերությունն այն է, որ հիմա արժեքը **կարող ենք օգտագործել**։

```python
ani_average = average_of([9, 10, 8])

print(ani_average)
print(ani_average > 8)
print(ani_average * 2)
```

| | Տպող ֆունկցիա | Վերադարձնող ֆունկցիա |
|---|---|---|
| Ի՞նչ է անում | ցույց է տալիս էկրանին | տալիս է քեզ արժեքը |
| Կարո՞ղ ես օգտագործել | ոչ | այո |
| Կարո՞ղ ես տպել | արդեն տպված է | `print(average_of(...))` |

> **Կանոն, որը գործում է մինչև դասընթացի վերջը.**
> ֆունկցիան կա՛մ **հաշվում է** (և վերադարձնում), կա՛մ **տպում է**։ Ոչ թե երկուսը։
>
> 21-րդ օրը դա կդառնա երկու առանձին ֆայլ։

`return`-ից հետո ֆունկցիան **անմիջապես ավարտվում է**։

```python
def has_passed(grade):
    if grade >= 4:
        return True
    return False

print(has_passed(9))
print(has_passed(2))
```

Այստեղ `else` պետք չէ. եթե պայմանը ճիշտ է, `return True`-ն արդեն ավարտել է ֆունկցիան։

Բայց այս ֆունկցիան իրականում ավելի կարճ է գրվում, որովհետև `grade >= 4`-ը **արդեն**
`True` կամ `False` է — դա 7-րդ օրվա դասն էր։

```python
def has_passed(grade):
    return grade >= 4

print(has_passed(9))
print(has_passed(2))
```

## Բ մաս: Լռելյայն արժեք

Անցողիկ միավորը 4 է, բայց ուրիշ դպրոցում կարող է ուրիշ լինել։ Կարող ենք պարամետրին
**լռելյայն արժեք** տալ։

```python
PASS_MARK = 4

def has_passed(grade, pass_mark=PASS_MARK):
    return grade >= pass_mark

print(has_passed(5))
print(has_passed(5, 6))
```

Առաջին կանչում երկրորդ արժեքը չգրեցինք, ուստի օգտագործվեց `PASS_MARK`-ը՝ 4։
Երկրորդում գրեցինք 6, ուստի օգտագործվեց 6։

**Լռելյայն արժեք ունեցող պարամետրերը գրվում են վերջում։**

## Գ մաս: Քո գործիքների հավաքածուն

Հիմա գրում ենք չորս ֆունկցիա, որոնք **վաղը դառնալու են քո ծրագրի սիրտը**։

```python
PASS_MARK = 4


def average_of(grades):
    """Վերադարձնում է գնահատականների միջինը՝ մեկ նիշով կլորացրած։"""
    total = 0
    for grade in grades:
        total = total + grade
    return round(total / len(grades), 1)


def has_passed(grade, pass_mark=PASS_MARK):
    """Վերադարձնում է True, եթե գնահատականը անցողիկ է։"""
    return grade >= pass_mark


def highest_of(class_grades):
    """Վերադարձնում է ամենաբարձր գնահատական ունեցող աշակերտի անունը։"""
    best_name = ""
    best_grade = -1
    for student_name, grade in class_grades.items():
        if grade > best_grade:
            best_grade = grade
            best_name = student_name
    return best_name


def failing_students(class_grades, pass_mark=PASS_MARK):
    """Վերադարձնում է չանցած աշակերտների անունների ցուցակը։"""
    failing = []
    for student_name, grade in class_grades.items():
        if grade < pass_mark:
            failing.append(student_name)
    return failing
```

> Երեք չակերտի մեջ գրված տողը ֆունկցիայի **նկարագրությունն** է։ Այն պարտադիր չէ,
> բայց երեք շաբաթ հետո ինքդ քեզ շնորհակալ կլինես։

Հիմա ստուգենք դրանք։

```python
class_grades = {"Անի": 9, "Դավիթ": 6, "Նարե": 10, "Արամ": 3,
                "Մարիամ": 8, "Տիգրան": 5, "Լիլիթ": 8, "Գոռ": 4,
                "Անահիտ": 9, "Հայկ": 2, "Սոնա": 8, "Վահե": 6}

print("Միջինը՝", average_of(list(class_grades.values())))
print("Ամենաբարձրը՝", highest_of(class_grades))
print("Չեն անցել՝", failing_students(class_grades))
print("Անին անցե՞լ է՝", has_passed(class_grades["Անի"]))
```

> `class_grades.values()` — տալիս է միայն արժեքները՝ առանց անունների։ `list(...)`-ը
> դրանք դարձնում է ցուցակ։ Սա միակ նոր բանն է այս բջիջում, և այն հուշաթերթում է։

Եվ հիմա՝ ամբողջ մատյանը, չորս ֆունկցիայով։

```python
print("=== ՄԱՏՅԱՆ ===")

for student_name, grade in class_grades.items():
    if has_passed(grade):
        result = "անցավ"
    else:
        result = "չանցավ"
    print(f"{student_name:<10} {grade:>3}  {result}")

print()
print(f"Միջինը՝ {average_of(list(class_grades.values()))}")
print(f"Ամենաբարձրը՝ {highest_of(class_grades)}")
print(f"Չեն անցել՝ {len(failing_students(class_grades))} աշակերտ")
```

```python
# Առաջադրանք 1 — քո չորս ֆունկցիան, քո դասարանով
# Պահիր այս բջիջը՝ վաղը այն տեղափոխելու ենք ֆայլ։

PASS_MARK = 4


def average_of(grades):
    """Վերադարձնում է գնահատականների միջինը։"""
    total = 0
    for grade in grades:
        total = total + grade
    return round(total / len(grades), 1)


def has_passed(grade, pass_mark=PASS_MARK):
    """Վերադարձնում է True, եթե գնահատականը անցողիկ է։"""
    return grade >= pass_mark


my_class = {"Անի": 9, "Դավիթ": 6, "Նարե": 10}

print(average_of(list(my_class.values())))
print(has_passed(my_class["Դավիթ"]))
```
