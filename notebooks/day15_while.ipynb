## Ա մաս: `while`

`for`-ը կապված է ցուցակի հետ։ Իսկ երբ ուզում ենք կրկնել, մինչև ինչ-որ բան **տեղի
ունենա**, պետք է `while`։

````python
count = 0

while count < 3:
    print("Պտույտ համար", count)
    count = count + 1

print("Վերջ")
````

`while պայման:` — «քանի դեռ պայմանը ճիշտ է, կրկնի՛ր»։

Ամեն պտույտից առաջ Python-ը ստուգում է պայմանը։ Երբ այն դառնում է `False`, ցիկլը կանգ
է առնում։

**Ուշադրություն `count = count + 1` տողին։** Առանց դրա պայմանը երբեք չէր փոխվի և ցիկլը
երբեք չէր կանգնի։

Իրական օգտագործումը՝ ցիկլ, որը սպասում է օգտվողին։ Դրա համար կա երկու բառ՝
`while True` և `break`։

````python
class_grades = []

while True:
    answer = input("Գնահատական (կամ «դուրս»)՝ ")

    if answer == "դուրս":
        break

    class_grades.append(int(answer))

print("Հավաքվեց՝", class_grades)
````

- `while True:` — «կրկնի՛ր միշտ»։ Պայմանը բառացիորեն `True` է, ուստի երբեք չի ավարտվի ինքն իրեն։
- `break` — «դուրս արի ցիկլից հիմա»։ Դա միակ ելքն է։

**Այս ձևը՝ `while True` + `break` — ամբողջ դասընթացում օգտագործում ենք միայն այսպես։**
Ուրիշ տարբերակներ կան, բայց պետք չեն։

## Բ մաս: Ստուգել, թե ինչ է գրել օգտվողը

4-րդ օրը տեսանք, որ `int("ինը")`-ը տալիս է `ValueError` և ծրագիրը կանգնում է։

Իսկական ծրագիրը չպետք է կանգնի այդ պատճառով։ Python-ում կա պատրաստի ստուգում.

````python
print("9".isdigit())
print("ինը".isdigit())
print("".isdigit())
````

`.isdigit()` — «արդյո՞ք սա միայն թվանշաններից է բաղկացած»։ Տալիս է `True` կամ `False`,
ուստի կարող է `if`-ի մեջ մտնել։

````python
class_grades = []

while True:
    answer = input("Գնահատական 1-ից 10 (կամ «դուրս»)՝ ")

    if answer == "դուրս":
        break

    if not answer.isdigit():
        print("Դա թիվ չէ։ Փորձի՛ր նորից։")
        continue

    grade = int(answer)

    if grade < 1 or grade > 10:
        print("Գնահատականը պետք է լինի 1-ից 10։ Փորձի՛ր նորից։")
        continue

    class_grades.append(grade)
    print(f"Ավելացվեց։ Հիմա {len(class_grades)} գնահատական է։")

print("Վերջնական ցուցակը՝", class_grades)
````

Նոր բառ՝ `continue` — «բաց թող մնացածը և անցի՛ր հաջորդ պտույտին»։

`break`-ը դուրս է գալիս ցիկլից։ `continue`-ն մնում է ցիկլում, բայց սկսում է նորից։

Ուշադրություն երեք ստուգման հերթականությանը.

1. Ուզո՞ւմ է դուրս գալ։
2. Թի՞վ է ընդհանրապես։
3. Ճիշտ միջակայքո՞ւմ է։

**Ամեն ստուգում ունի իր առանձին հաղորդագրությունը։** «Սխալ մուտք» ոչինչ չի ասում։
«Գնահատականը պետք է լինի 1-ից 10» ասում է, թե ինչ անել։

## Գ մաս: Անվերջ ցիկլ

Ամենահաճախ հանդիպող սխալը `while`-ի հետ՝ պայմանը երբեք չի փոխվում։

```python
count = 0

while count < 3:
    print("Բարև")        # count-ը չի փոխվում
```

Այս կոդը կտպի «Բարև» **անվերջ**։

**Մի՛ գործարկիր այն։** Փոխարենը իմացիր, թե ինչպես կանգնեցնել.

<b>4. Պարտադիր։</b> Ավելացրո՛ւ ստուգում, որ գնահատականը 1-ից 10 է։<br/><br/>
<b>5. Ցանկության դեպքում։</b> Ավելացրո՛ւ «ջնջել» հրաման, որը հանում է վերջին
գնահատականը։
</p>

````python
# Առաջադրանք 1, 2, 3 — հավաքել գնահատականներ մինչև «դուրս»

class_grades = []

while True:
    answer = input("Գնահատական (կամ «դուրս»)՝ ")

    if answer == "դուրս":
        break

    if not answer.isdigit():
        print("Դա թիվ չէ։")
        continue

    class_grades.append(int(answer))

print(f"Հավաքվեց {len(class_grades)} գնահատական")
````

## Ինչի հասանք

- `while պայման:` — կրկնում է, քանի դեռ պայմանը ճիշտ է։
- **`while True:` + `break`** — սա է այն ձևը, որ օգտագործում ենք։
- `continue` — բաց թողնել մնացածը և անցնել հաջորդ պտույտին։
- `.isdigit()` — թի՞վ է տեքստը։ Ստուգում ենք **նախքան** `int()`-ը։
- Անվերջ ցիկլը կանգնեցվում է **■** կոճակով։ Համակարգիչը չի կոտրվել։
