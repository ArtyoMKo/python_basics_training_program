## Ա մաս: Կուտակիչ

Խնդիրը՝ գումարել բոլոր գնահատականները։

Ձեռքով դա այսպես կլիներ.

```python
class_grades = [9, 6, 10, 3, 8]

print(9 + 6 + 10 + 3 + 8)
```

Աշխատում է հինգի համար։ Քառասունի համար՝ ոչ։

Գաղափարը՝ պահել «մինչ այժմ հավաքած գումարը» մի փոփոխականում և ամեն պտույտում
ավելացնել նրան։

```python
class_grades = [9, 6, 10, 3, 8]

total = 0

for grade in class_grades:
    total = total + grade

print(total)
```

Երեք մասից բաղկացած ձև, որը կհանդիպի ամբողջ դասընթացում.

1. **Մինչ ցիկլը՝** `total = 0` — սկսում ենք զրոյից
2. **Ցիկլի ներսում՝** `total = total + grade` — ավելացնում ենք
3. **Ցիկլից հետո՝** `print(total)` — նայում ենք արդյունքին

Ուշադրություն՝ `print`-ը **ցիկլից դուրս** է։ Եթե ներսում լիներ, կտպեր ամեն պտույտում։

Տեսնենք, թե ինչպես է այն աճում։

```python
class_grades = [9, 6, 10, 3, 8]

total = 0

for grade in class_grades:
    total = total + grade
    print("Ավելացրի", grade, "— հիմա գումարը՝", total)

print("Վերջնական գումարը՝", total)
```

Նույն ձևով կարելի է **հաշվել**, ոչ միայն գումարել։ Ավելացնում ենք 1՝ գնահատականի փոխարեն։

```python
class_grades = [9, 6, 10, 3, 8]

how_many = 0

for grade in class_grades:
    how_many = how_many + 1

print(how_many)
```

Այստեղ `len()`-ը ավելի կարճ կլիներ։ Բայց հաշվիչի ձևը պետք կգա **վաղը**, երբ ուզենք
հաշվել ոչ թե բոլորին, այլ միայն նրանց, ովքեր չեն անցել։

## Բ մաս: Միջինը

Միջինը գումարն է՝ բաժանած քանակի։ Երկուսն էլ արդեն ունենք։

```python
class_grades = [9, 6, 10, 3, 8, 5, 8, 4, 9, 2, 8, 6]

total = 0

for grade in class_grades:
    total = total + grade

class_average = total / len(class_grades)

print("Գումարը՝", total)
print("Աշակերտների թիվը՝", len(class_grades))
print("Միջինը՝", class_average)
```

`6.5` — և ահա, 3-րդ օրվա `float`-ը վերջապես իմաստ ունի։

Տասներկու գնահատականի գումարը 78 է, բաժանած 12-ի՝ 6.5։ **Միջինը գրեթե երբեք ամբողջ
թիվ չէ**, և դա հենց այն դեպքն է, որի համար տասնորդական թվեր կան։

Երբեմն միջինը երկար է ստացվում։

```python
class_grades = [9, 6, 10, 3, 8, 5, 8]

total = 0
for grade in class_grades:
    total = total + grade

print(total / len(class_grades))
```

Մատյանում այդպիսի թիվ չես գրի։ Կլորացնում ենք՝ 4-րդ օրվա `round`-ով։

```python
class_average = total / len(class_grades)

print(round(class_average, 1))
print(f"Դասարանի միջինը՝ {class_average:.1f}")
```

Երկու ձևն էլ ճիշտ է։ `round()` — երբ թիվը պետք է հետագայում օգտագործես։
`:.1f` — երբ միայն տպում ես։

## Գ մաս: `range` — թվերի շարք

Երբեմն պետք է կրկնել որոշակի **թվով** անգամ, ոչ թե ցուցակի վրայով։

```python
for number in range(1, 11):
    print(number)
```

`range(1, 11)` — 1-ից **մինչև** 11, առանց 11-ի։ Վերջին թիվը ներառված չէ։

Սա սկզբում տարօրինակ է թվում, բայց նույն տրամաբանությունն է, ինչ երեկվա `[0:3]`-ը։

Եթե միայն մեկ թիվ գրես, սկսում է զրոյից։

```python
for number in range(5):
    print(number)
```

Դասարանի համար դա օգտակար է գնահատականների բաշխումը հաշվելու համար։

```python
class_grades = [9, 6, 10, 3, 8, 5, 8, 4, 9, 2, 8, 6]

for mark in range(1, 11):
    how_many = 0
    for grade in class_grades:
        if grade == mark:
            how_many = how_many + 1
    print(f"{mark} միավոր՝ {how_many} աշակերտ")
```

Այստեղ **ցիկլ ցիկլի ներսում** է — երկու անգամ ներս տեղաշարժ։ Դա միակ տեղն է դասընթացում,
որտեղ դա անում ենք։ Կարդա՛ այն դանդաղ, բայց անգիր անելու կարիք չկա։

## Իսկ հիմա՝ նոր խնդիր

Դասարանի միջինը գիտենք։ Բայց փորձենք պատասխանել՝ **ո՞ւմ գնահատականն է 10-ը**։

```python
student_names = ["Անի", "Դավիթ", "Նարե", "Արամ", "Մարիամ"]
class_grades = [9, 6, 10, 3, 8]

for position in range(len(class_grades)):
    if class_grades[position] == 10:
        print(student_names[position])
```

Աշխատում է — բայց ամեն ինչ կախված է նրանից, որ **երկու ցուցակը նույն հերթականությամբ
լինեն**։

Եթե ինչ-որ մեկը `student_names`-ից աշակերտ հեռացնի, իսկ `class_grades`-ից՝ ոչ, ապա
բոլոր գնահատականները կտեղաշարժվեն մեկով, և **ամեն ինչ սխալ կլինի առանց որևէ սխալի
հաղորդագրության**։

Սա վտանգավոր է։ **Լուծվում է 16-րդ օրը։**

<b>1. Պարտադիր։</b> Հաշվի՛ր քո դասարանի գնահատականների գումարը ցիկլով։<br/><br/>
<b>2. Պարտադիր։</b> Հաշվի՛ր և տպի՛ր <b>քո դասարանի իրական միջինը</b>՝ մեկ նիշով

<b>3. Պարտադիր։</b> Տպի՛ր 1-ից 10 թվերը <code>range</code>-ով։<br/><br/>
<b>4. Պարտադիր։</b> Հաշվի՛ր, թե քո դասարանում քանի աշակերտ ունի 10 միավոր։<br/><br/>
<b>5. Ցանկության դեպքում։</b> Տպի՛ր գնահատականների բաշխումը՝ 1-ից 10, ամեն մեկի դիմաց՝
քանի աշակերտ։

```python
# Առաջադրանք 1 և 2 — գումարը և միջինը

class_grades = [9, 6, 10, 3, 8]

total = 0
for grade in class_grades:
    total = total + grade

print(f"Գումարը՝ {total}")
print(f"Միջինը՝ {total / len(class_grades):.1f}")
```

```python
# Առաջադրանք 4 — քանի՞ աշակերտ ունի 10 միավոր

how_many = 0

for grade in class_grades:
    if grade == 10:
        how_many = how_many + 1

print(how_many)
```

## Ինչի հասանք

- **Կուտակիչի ձևը՝** `total = 0` մինչ ցիկլը, `total = total + ...` ներսում,
  `print` հետո։
- Միջինը՝ `total / len(...)`։ Գրեթե միշտ տասնորդական թիվ է։
- `round(x, 1)` կամ `f"{x:.1f}"` — կլորացնում են։
- `range(1, 11)` — 1-ից 10։ **Վերջին թիվը ներառված չէ։**

**հերթականությամբ**, և դա փխրուն է. մեկ սխալ տեղաշարժ, և ամեն ինչ սխալ է առանց սխալի
