## Ա մաս: `for`

Ահա երեկվա վերջին բջիջը։

````python
student_names = ["Անի", "Դավիթ", "Նարե", "Արամ", "Մարիամ"]

print(student_names[0])
print(student_names[1])
print(student_names[2])
print(student_names[3])
print(student_names[4])
````

Եվ ահա նույնը։

````python
student_names = ["Անի", "Դավիթ", "Նարե", "Արամ", "Մարիամ"]

for student_name in student_names:
    print(student_name)
````

**Երկու տող։** Եվ ամենակարևորը՝ ցուցակի երկարությունը նշանակություն չունի։

Ապացուցենք. ավելացնում ենք տասնհինգ աշակերտ և **ոչ մի տող չենք փոխում**։

````python
student_names = ["Անի", "Դավիթ", "Նարե", "Արամ", "Մարիամ", "Տիգրան", "Լիլիթ", "Գոռ",
                 "Անահիտ", "Հայկ", "Սոնա", "Վահե", "Մանե", "Սամվել", "Էլեն", "Նարեկ",
                 "Ալիս", "Արթուր", "Սիրանուշ", "Վարդան"]

for student_name in student_names:
    print(student_name)
````

Քսան աշակերտ, նույն երկու տողը։ Երեք հարյուր աշակերտ — նույն երկու տողը։

## Բ մաս: Ի՞նչ է կատարվում ամեն պտույտում

Կարդանք բառ առ բառ.

```
for student_name in student_names:
    print(student_name)
```

- `for` — «յուրաքանչյուրի համար»
- `student_name` — **անուն, որ մենք ենք ընտրում**։ Ամեն պտույտում այն ստանում է
  հաջորդ արժեքը
- `in student_names` — որ ցուցակից
- `:` և չորս բացատ — ուղիղ ինչպես `if`-ում

Ցիկլի անունը մերն է։ Կարող էինք գրել `x`, բայց `student_name`-ը կարդալի է։

Տեսնենք, թե ինչ է լինում փոփոխականի հետ ամեն պտույտում։

````python
student_names = ["Անի", "Դավիթ", "Նարե"]

for student_name in student_names:
    print("Այս պտույտում student_name-ը հավասար է՝", student_name)
````

Երեք պտույտ, երեք տարբեր արժեք։ Ոչ մի կախարդանք — պարզապես Python-ը կրկնում է
ներսի տողերը՝ ամեն անգամ նոր արժեքով։

Ներսում կարող է լինել **որքան ուզես կոդ**, ոչ միայն մեկ տող։

````python
student_names = ["Անի", "Դավիթ", "Նարե"]

for student_name in student_names:
    print("Աշակերտ՝", student_name)
    print("Ներկա է")
    print()
````

Իսկ ցիկլից **դուրս** գրվածը կատարվում է մեկ անգամ։ Բացատներն են որոշում։

````python
student_names = ["Անի", "Դավիթ", "Նարե"]

print("=== Դասարանի ցուցակ ===")

for student_name in student_names:
    print(student_name)

print("=== Վերջ ===")
````

## Գ մաս: Ցիկլը և պայմանը միասին

Ցիկլի ներսում կարող է լինել `if`։ Այդ դեպքում **երկու անգամ** ենք ներս տեղաշարժում։

````python
class_grades = [9, 6, 10, 3, 8]
PASS_MARK = 4

for grade in class_grades:
    if grade >= PASS_MARK:
        print(grade, "— անցավ")
    else:
        print(grade, "— չանցավ")
````

**Կանգ առ և նայի՛ր այս բջիջին։**

9-րդ օրը այս նույն բանը արեցինք։ Այն զբաղեցնում էր **քսանհինգ բլոկ**՝ ավելի քան հարյուր
տող։

Հիմա այն **հինգ տող** է, և աշխատում է ցանկացած թվով աշակերտի համար։

## Երկու ցուցակ միասին

Մեկ խնդիր մնաց. ցիկլը տալիս է **կա՛մ** անունը, **կա՛մ** գնահատականը։ Իսկ մեզ երկուսն
էլ պետք են։

Առայժմ լուծենք ինդեքսով։

````python
student_names = ["Անի", "Դավիթ", "Նարե"]
class_grades = [9, 6, 10]

for position in range(len(student_names)):
    print(f"{student_names[position]}: {class_grades[position]}")
````

`range(len(...))` — տալիս է 0, 1, 2 ... մինչև ցուցակի վերջը։ Վաղը `range`-ին առանձին
կնայենք։

**Աշխատում է, բայց տգեղ է։** Երկու ցուցակ, որոնք պետք է միշտ նույն հերթականությամբ
լինեն։ Եթե մեկում աշակերտ ավելացնես, իսկ մյուսում՝ ոչ, ամեն ինչ կխառնվի, և
**ոչ մի սխալ չես տեսնի**։

Այդ խնդիրը լուծվում է **16-րդ օրը**՝ բառարանով։

````python
# Առաջադրանք 1 — բոլոր անունները ցիկլով

student_names = ["Անի", "Դավիթ", "Նարե"]

for student_name in student_names:
    print(student_name)
````

````python
# Առաջադրանք 3 — ցիկլ, որի ներսում պայման է

class_grades = [9, 6, 10, 3, 8]
PASS_MARK = 4

for grade in class_grades:
    if grade >= PASS_MARK:
        print(grade, "— անցավ")
    else:
        print(grade, "— չանցավ")
````

## Ինչի հասանք

- `for արժեք in ցուցակ:` — կրկնում է ներսի կոդը ամեն տարրի համար։
- **Ցուցակի երկարությունը նշանակություն չունի։** Նույն կոդը՝ 5 և 500 աշակերտի համար։
- Ցիկլի ներսում կարող է լինել `if`։ Երկու անգամ ներս տեղաշարժ։
