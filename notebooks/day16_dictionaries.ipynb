## Ա մաս: Երկու ցուցակի խնդիրը

Այսպես ենք աշխատել մինչ այժմ։

````python
student_names = ["Անի", "Դավիթ", "Նարե", "Արամ"]
class_grades = [9, 6, 10, 3]

print(f"{student_names[2]}-ի գնահատականը՝ {class_grades[2]}")
````

Հիմա պատկերացրու, որ Դավիթը տեղափոխվեց ուրիշ դպրոց և դու հանում ես նրան ցուցակից —
**բայց մոռանում ես հանել գնահատականը**։

````python
student_names = ["Անի", "Նարե", "Արամ"]
class_grades = [9, 6, 10, 3]

print(f"{student_names[2]}-ի գնահատականը՝ {class_grades[2]}")
````

**Արամը հանկարծ ստացավ 10։**

Ոչ մի կարմիր տեքստ։ Ոչ մի սխալի հաղորդագրություն։ Ծրագիրը հանգիստ աշխատեց և տվեց
սխալ պատասխան։ Իսկ եթե դա մատյան լիներ, ոչ ոք չէր նկատի։

**Խնդիրը արմատային է.** անունն ու գնահատականը կապված են միայն նրանով, որ
պատահաբար նույն տեղում են։

## Բ մաս: Բառարան

Բառարանում արժեքը պահվում է **անվան տակ**, ոչ թե տեղում։

````python
class_grades = {"Անի": 9, "Դավիթ": 6, "Նարե": 10, "Արամ": 3}

print(class_grades)
````

Ձևավոր փակագծեր `{ }`, և ամեն զույգը՝ `բանալի: արժեք`։

- **Բանալին** (`"Անի"`) — այն, ինչով փնտրում ենք
- **Արժեքը** (`9`) — այն, ինչ գտնում ենք

Հասնում ենք արժեքին ուղիղ այնպես, ինչպես ցուցակում — բայց թվի փոխարեն գրում ենք
**անունը**։

````python
print(class_grades["Անի"])
print(class_grades["Նարե"])
````

**Եվ հիմա Դավիթին հանելը անվտանգ է։**

````python
class_grades = {"Անի": 9, "Նարե": 10, "Արամ": 3}

print(f"Արամի գնահատականը՝ {class_grades['Արամ']}")
````

Արամը ստացավ իր 3-ը։ Հերթականությունը նշանակություն չունի, որովհետև
**գնահատականը կապված է անվան հետ, ոչ թե տեղի**։

> Ուշադրություն f-տողի ներսում չակերտներին. դրսում կրկնակի են (`f"..."`), ուստի
> ներսում գրում ենք միակի (`'Արամ'`)։ Հակառակը նույնպես աշխատում է։

## Գ մաս: Ավելացնել, ուղղել, ստուգել

Նոր աշակերտ ավելացնելը՝ ուղղակի գրում ես նոր բանալի։

````python
class_grades = {"Անի": 9, "Նարե": 10, "Արամ": 3}

class_grades["Գոռ"] = 4

print(class_grades)
````

Գնահատականը ուղղելը՝ **ուղիղ նույն տողը**։

````python
class_grades["Անի"] = 10

print(class_grades)
````

Եթե բանալին չկա՝ ավելանում է։ Եթե կա՝ փոխարինվում է։ Մեկ ձև՝ երկու գործողության համար։

Հանելը՝ `del`։

````python
del class_grades["Արամ"]

print(class_grades)
````

Իսկ քանիսն են՝ `len()`, ուղիղ ինչպես ցուցակում։

````python
print(len(class_grades))
````

### Հիմա կոտրենք

Ի՞նչ է լինում, երբ փնտրում ես աշակերտի, ով բառարանում չկա։

````python
print(class_grades["Արամ"])
````

### Ինչ գրված է այնտեղ

```
KeyError: 'Արամ'
```

«Այդ բանալին բառարանում չկա»։ Python-ը ուղիղ ցույց է տալիս, թե որն է չգտել։

Երկու սովորական պատճառ.

1. **Տառասխալ** — «Անի» և «անի» տարբեր բանալիներ են։
2. **Աշակերտը իսկապես չկա** — հենց նոր հանեցինք Արամին։

Խուսափելու ձևը՝ `in`, ուղիղ ինչպես ցուցակում։

````python
if "Արամ" in class_grades:
    print(class_grades["Արամ"])
else:
    print("Արամը այս դասարանում չէ")
````

<b>3. Պարտադիր։</b> Ավելացրո՛ւ նոր աշակերտ, ուղղի՛ր մեկի գնահատականը, հանի՛ր մեկին։

<code>KeyError</code>-ը, հետո ուղղի՛ր <code>in</code>-ով։<br/><br/>
<b>5. Ցանկության դեպքում։</b> Գրի՛ր ստուգում, որը հարցնում է անուն և տպում գնահատականը,
կամ ասում է, որ այդպիսի աշակերտ չկա։

````python
# Առաջադրանք 1 — քո դասարանը մեկ բառարանով

class_grades = {"Անի": 9, "Դավիթ": 6}

print(class_grades)
````

````python
# Առաջադրանք 3 — ավելացնել, ուղղել, հանել

class_grades["Նարե"] = 10
class_grades["Անի"] = 10
del class_grades["Դավիթ"]

print(class_grades)
````

````python
# Առաջադրանք 4 — կոտրիր դիտավորյալ, հետո ուղղիր in-ով

if "Դավիթ" in class_grades:
    print(class_grades["Դավիթ"])
else:
    print("Այդպիսի աշակերտ չկա")
````

## Ինչի հասանք

- `{"Անի": 9}` — **բանալի → արժեք**։ Անունը ներս, գնահատականը դուրս։
- `class_grades["Անի"] = 10` — ավելացնում է, եթե չկա, ուղղում է, եթե կա։
- `del` — հանում է։ `len()` — քանիսն են։ `in` — կա՞։
- `KeyError` — այդ բանալին չկա։ Խուսափում ենք `in`-ով։
- **Հերթականությունը այլևս նշանակություն չունի։** 13-րդ օրվա խնդիրը՝ լուծված։
