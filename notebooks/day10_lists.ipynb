## Բ մաս: Ինչպես հասնել մեկ արժեքի

````python
class_grades = [9, 6, 10, 3, 8]

print(class_grades[0])
````

Քառակուսի փակագծերի մեջ գրում ենք **տեղը** — ինդեքսը։

Եվ ահա այսօրվա միակ շփոթեցնող բանը.

> **Հաշվարկը սկսվում է զրոյից։**

````python
class_grades = [9, 6, 10, 3, 8]

print(class_grades[0])     # առաջինը
print(class_grades[1])     # երկրորդը
print(class_grades[4])     # հինգերորդը
````

| Ինդեքսը | Որերորդն է | Արժեքը |
|---|---|---|
| `[0]` | առաջինը | 9 |
| `[1]` | երկրորդը | 6 |
| `[2]` | երրորդը | 10 |
| `[3]` | չորրորդը | 3 |
| `[4]` | հինգերորդը | 8 |

Հինգ տարր, ինդեքսները՝ **0-ից 4**։ Ոչ թե 1-ից 5։

Վերջինին հասնելու համար կա հարմարություն՝ `-1`։

````python
print(class_grades[-1])
````

Երկու ցուցակը միասին օգտագործելով՝ կարող ենք տպել ամբողջական տող։

````python
student_names = ["Անի", "Դավիթ", "Նարե"]
class_grades = [9, 6, 10]

print(f"{student_names[0]}: {class_grades[0]}")
print(f"{student_names[1]}: {class_grades[1]}")
print(f"{student_names[2]}: {class_grades[2]}")
````

Աշխատում է — բայց նայի՛ր ուշադիր։ Երեք աշակերտի համար՝ երեք տող։ Քառասունի համար՝
քառասուն տող։

**Այդ խնդիրը վաղը չէ մյուս օրը լուծվում է։** Այսօր ուշադրություն դարձրու միայն նրան,
որ խնդիրը կա։

## Գ մաս: Հիմա կոտրենք

Հինգ տարր ունեցող ցուցակում ի՞նչ կա `[5]` տեղում։

**Նախ գուշակի՛ր։**

````python
class_grades = [9, 6, 10, 3, 8]

print(class_grades[5])
````

### Ինչ գրված է այնտեղ

```
IndexError: list index out of range
```

«Այդ տեղը ցուցակում չկա»։

Հինգ տարր՝ ինդեքսները 0, 1, 2, 3, 4։ `[5]`-ը **վեցերորդն** է, իսկ վեցերորդ չկա։

> **Կանոն, որը արժե հիշել.** եթե ցուցակում `n` տարր կա, վերջինի ինդեքսը `n - 1` է։
> Կամ պարզապես գրի՛ր `[-1]` և չմտածես։

Սա ամենահաճախ հանդիպող սխալն է ցուցակների հետ աշխատելիս, և գրեթե միշտ պատճառը մեկն է՝
մոռացել ես, որ զրոյից է հաշվում։
