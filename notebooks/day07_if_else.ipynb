## Ա մաս: Համեմատել

3-րդ օրը տեսանք `True` և `False` արժեքները և ասացինք, որ դրանք առայժմ անօգուտ են։
Այսօր դրանք դառնում են ամենակարևորը։

````python
print(9 > 4)
print(2 > 4)
````

Համեմատությունը արժեք է — ուղիղ այնպես, ինչպես `9`-ը կամ `"Անի"`-ն։ Այն կամ `True` է,
կամ `False`։

````python
grade = 9

print(grade > 4)
print(type(grade > 4))
````

Վեց նշան, և ամբողջ դասընթացում ուրիշ բան պետք չի գա։

````python
print(9 > 4)      # մեծ է
print(9 < 4)      # փոքր է
print(9 >= 9)     # մեծ կամ հավասար
print(9 <= 4)     # փոքր կամ հավասար
print(9 == 9)     # հավասար է
print(9 != 4)     # հավասար չէ
````

> **Ուշադրություն՝ `==` երկու նշան։**
>
> `=` (մեկ նշան) — **վերագրում է**. `grade = 9` նշանակում է «դիր 9-ը `grade`-ի մեջ»։
> `==` (երկու նշան) — **հարցնում է**. `grade == 9` նշանակում է «արդյո՞ք 9 է»։
>
> Սա ամենահաճախ շփոթվող բանն է ծրագրավորման սկզբում։ Երբ ինչ-որ բան տարօրինակ է
> աշխատում, ստուգի՛ր, թե քանի նշան ես գրել։

Համեմատել կարելի է նաև տեքստ։

````python
print("Անի" == "Անի")
print("Անի" == "անի")
````

Մեծատառը և փոքրատառը տարբեր տառեր են։ Սա 22-րդ օրը կարևոր կդառնա, երբ ֆայլից ենք
անուններ կարդալու։

## Բ մաս: `if` և `else`

Հիմա այդ `True`/`False`-ը օգտագործում ենք որոշում ընդունելու համար։

````python
PASS_MARK = 4
grade = 9

if grade >= PASS_MARK:
    print("անցավ")
````

Կարդանք բառ առ բառ.

- `if` — «եթե»
- `grade >= PASS_MARK` — պայմանը։ Այն կամ `True` է, կամ `False`
- `:` — երկու կետ։ **Պարտադիր է։** Նշանակում է «ահա թե ինչ անել»
- չորս բացատ — ասում է, թե որ տողերն են պայմանի ներսում

Փոխի՛ր `grade`-ը 2-ի և գործարկի՛ր նորից։ Ոչինչ չի տպվի — և դա ճիշտ է։

Սովորաբար ուզում ենք, որ մյուս դեպքում **ուրիշ** բան լինի։ Դրա համար կա `else`։

````python
PASS_MARK = 4
grade = 2

if grade >= PASS_MARK:
    print("անցավ")
else:
    print("չանցավ")
````

`else`-ը պայման չունի։ Այն նշանակում է պարզապես «մնացած բոլոր դեպքերում»։

Հիմա ամբողջական տող՝ f-տողի հետ միասին։

````python
PASS_MARK = 4
student_name = "Հայկ"
grade = 2

if grade >= PASS_MARK:
    print(f"{student_name}: {grade} — անցավ")
else:
    print(f"{student_name}: {grade} — չանցավ")
````

## Գ մաս: Չորս բացատը

Այն, ինչ Word-ում ձևավորում է, Python-ում **իմաստ** է։

````python
grade = 2

if grade >= 4:
    print("Այս տողը պայմանի ներսում է")
print("Իսկ այս տողը՝ ոչ")
````

Երկրորդ տողը տպվեց, չնայած պայմանը `False` էր։ Որովհետև այն **ներսում չէ** —
բացատներ չունի։

Բացատները ցույց են տալիս, թե ինչն է ինչի ներսում։ Դա ամբողջ կանոնը։

````python
# Առաջադրանք 1 — քո դպրոցի անցողիկ միավորը

PASS_MARK = 4
````

````python
# Առաջադրանք 2 — երեք աշակերտ։ Առաջինը գրված է։

student_name = "Անի"
grade = 9

if grade >= PASS_MARK:
    print(f"{student_name}: անցավ")
else:
    print(f"{student_name}: չանցավ")
````

````python
# Առաջադրանք 4 — կոտրիր դիտավորյալ՝ ջնջելով բացատները, հետո ուղղիր

if 9 >= PASS_MARK:
    print("անցավ")
````

## Ինչի հասանք

- Համեմատությունը արժեք է՝ `True` կամ `False`։
- `=` վերագրում է, `==` հարցնում է։ Երկու տարբեր բան։
- `if պայման:` ... `else:` — ծրագիրը որոշում է ընդունում։
- **Չորս բացատը իմաստ ունի**, ոչ թե գեղեցկություն։
