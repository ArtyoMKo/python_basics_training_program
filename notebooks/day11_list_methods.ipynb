Դասարանը փոխվում է. աշակերտ է գալիս, աշակերտ է հեռանում, ցուցակը պետք է այբբենական
կարգով։ Այսօր այդ ամենի մասին է։

## ԱՅՍՕՐ:

- **Ա մաս:** ավելացնել և հեռացնել
- **Բ մաս:** կա՞ ցուցակում
- **Գ մաս:** դասավորել և կտրել

<h3 style="color:#06c; margin-top:0;">🔄 Որտեղ էինք մնացել</h3>

<code>class_grades = [9, 6, 10]</code> — մեկ անուն, շատ արժեք։<br/>
<code>len()</code> — քանիսն են։ <code>[0]</code> — առաջինը։ <code>[-1]</code> — վերջինը։

```python
student_names = ["Անի", "Դավիթ", "Նարե", "Արամ", "Մարիամ"]

print(student_names)
```

## Ա մաս: Ավելացնել և հեռացնել

Նոր աշակերտ եկավ դասարան։

```python
student_names.append("Տիգրան")

print(student_names)
```

`.append(...)` — ավելացնում է **վերջում**։

Ուշադրություն կետին. `student_names.append(...)` կարդացվում է որպես
«`student_names`-ին ասա՝ ավելացրու»։ Այս ձևը՝ անուն, կետ, հրահանգ — շատ կհանդիպի։

Եվ ուշադրություն ևս մեկ բանի. բջիջը ոչինչ չտպեց, բայց **ցուցակը փոխվեց**։

```python
student_names.remove("Արամ")

print(student_names)
```

`.remove(...)` — հեռացնում է **արժեքով**, ոչ թե տեղով։ Գրում ես, թե ում ես հեռացնում,
ոչ թե որերորդին։

## Բ մաս: Կա՞ ցուցակում

```python
print("Անի" in student_names)
print("Արամ" in student_names)
```

`in` — տալիս է `True` կամ `False`։ Իսկ դա նշանակում է, որ այն կարող է պայման դառնալ։

```python
if "Անի" in student_names:
    print("Անին այս դասարանում է")
else:
    print("Անին այս դասարանում չէ")
```

Սա շատ օգտակար է `remove`-ից առաջ։ Եթե փորձես հեռացնել մեկին, ով ցուցակում չկա,
ծրագիրը կկանգնի սխալով։

```python
if "Վահե" in student_names:
    student_names.remove("Վահե")
else:
    print("Վահեն այս դասարանում չէ — ոչինչ չհեռացրի")
```

## Գ մաս: Դասավորել և կտրել

```python
student_names.sort()

print(student_names)
```

`.sort()` — դասավորում է **տեղում**։ Ցուցակը ինքն է փոխվում, նոր ցուցակ չի ստեղծվում։

Հայերեն անունները դասավորվում են այբբենական կարգով ճիշտ։

Թվերի համար՝ նույնը, փոքրից մեծ։

```python
class_grades = [9, 6, 10, 3, 8]
class_grades.sort()

print(class_grades)
```

Իսկ եթե պետք է մեծից փոքր՝

```python
class_grades.sort(reverse=True)

print(class_grades)
```

**Կտրել** — վերցնել ցուցակի մի մասը։

```python
student_names = ["Անի", "Դավիթ", "Նարե", "Արամ", "Մարիամ"]

print(student_names[0:3])
```

`[0:3]` նշանակում է «սկսած 0-ից, **մինչև** 3-ը, առանց 3-ի»։

Սա նույն տրամաբանությունն է, ինչ `range`-ինը, որ կտեսնենք վաղը. **վերջին թիվը ներառված չէ**։

```python
print(student_names[0:2])     # առաջին երկուսը
print(student_names[2:5])     # երրորդից հինգերորդը
print(student_names[-2:])     # վերջին երկուսը
```

## Իսկ հիմա՝ խնդիրը

Ցուցակը հիմա հարմար է։ Բայց փորձենք տպել բոլոր անունները մեկ առ մեկ։

```python
student_names = ["Անի", "Դավիթ", "Նարե", "Արամ", "Մարիամ"]

print(student_names[0])
print(student_names[1])
print(student_names[2])
print(student_names[3])
print(student_names[4])
```

Հինգ աշակերտ՝ հինգ տող։

**Քառասուն աշակերտ՝ քառասուն տող։** Եվ եթե մեկը ավելանա, պետք է նոր տող գրել։
Եվ եթե մեկը հեռանա, `[4]`-ը կդառնա `IndexError`։

Ծանո՞թ է։ Սա ուղիղ 6-րդ օրվա խնդիրն է՝ նոր հագուստով։

աշակերտ <code>.append()</code>-ով։<br/><br/>
<b>2. Պարտադիր։</b> Հեռացրո՛ւ մեկին <code>.remove()</code>-ով։ <b>Նախ ստուգի՛ր
<code>in</code>-ով, որ նա ցուցակում է։</b><br/><br/>
<b>3. Պարտադիր։</b> Դասավորի՛ր անունները այբբենական կարգով և տպի՛ր։<br/><br/>
<b>4. Պարտադիր։</b> Տպի՛ր քո դասարանի <b>բոլոր</b> անունները՝ մեկ առ մեկ,
<code>[0]</code>, <code>[1]</code>, ... ձևով։ <b>Հաշվի՛ր, թե քանի տող գրեցիր։</b><br/><br/>
<b>5. Ցանկության դեպքում։</b> Տպի՛ր միայն առաջին երեքին՝ կտրելով։
</p>
</div>

```python
# Առաջադրանք 1 և 2 — ավելացնել և հեռացնել

student_names = ["Անի", "Դավիթ", "Նարե"]
student_names.append("Տիգրան")

print(student_names)
```

```python
# Առաջադրանք 3 — դասավորել

student_names.sort()

print(student_names)
```

```python
# Առաջադրանք 4 — բոլորը մեկ առ մեկ։ Քանի՞ տող գրեցիր։

print(student_names[0])
```

## Ինչի հասանք

- `.append(...)` — ավելացնում է վերջում։ `.remove(...)` — հեռացնում է արժեքով։
- `in` — կա՞ ցուցակում։ Պայման է, ուրեմն կարելի է `if`-ի մեջ դնել։
- `.sort()` — դասավորում է տեղում։ `reverse=True` — հակառակ կարգով։
- `[0:3]` — կտրում է մի մասը։ **Վերջին թիվը ներառված չէ։**

## Ինչն է դեռ մեզ նյարդայնացնում

Ցուցակը հարմար է։ Բայց **քառասուն անուն տպելու համար դեռ քառասուն տող է պետք**։

Ամեն անգամ նույն բանը՝ `print(student_names[ինչ-որ թիվ])`։ Փոխվում է միայն թիվը։
