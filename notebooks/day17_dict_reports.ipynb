Երեկ գտնում էինք **մեկ** աշակերտի գնահատականը։ Այսօր տպում ենք **ամբողջ մատյանը**։

## ԱՅՍՕՐ:

- **Ա մաս:** ցիկլ բառարանի վրայով
- **Բ մաս:** մի քանի գնահատական մեկ աշակերտի համար
- **Գ մաս:** ամբողջական հաշվետվություն

<div style="border-left: 6px solid #06c; background: #f2f7ff; padding: 12px 16px; margin: 12px 0;">
<h3 style="color:#06c; margin-top:0;">🔄 Որտեղ էինք մնացել</h3>

<code>class_grades = {"Անի": 9}</code> — բանալի → արժեք։ Հերթականությունը նշանակություն չունի։
</p>
</div>

````python
PASS_MARK = 4

class_grades = {"Անի": 9, "Դավիթ": 6, "Նարե": 10, "Արամ": 3,
                "Մարիամ": 8, "Տիգրան": 5, "Լիլիթ": 8, "Գոռ": 4,
                "Անահիտ": 9, "Հայկ": 2, "Սոնա": 8, "Վահե": 6}

print(len(class_grades), "աշակերտ")
````

## Ա մաս: Ցիկլ բառարանի վրայով

Բառարանի ամեն տարրը **երկու** բան է՝ անուն և գնահատական։ Ուստի ցիկլում երկու անուն ենք գրում։

````python
for student_name, grade in class_grades.items():
    print(f"{student_name}: {grade}")
````

Կարդանք.

- `.items()` — «տուր ինձ զույգերը»
- `student_name, grade` — **երկու անուն, ստորակետով**։ Առաջինը ստանում է բանալին,
  երկրորդը՝ արժեքը
- մնացածը՝ սովորական `for`, ինչպես 12-րդ օրը

Անունները մերն են։ Կարող էինք գրել `a, b`, բայց այդ դեպքում ոչինչ չէր կարդացվի։

Հիմա 14-րդ օրվա բոլոր առաջադրանքները դառնում են ավելի պարզ։ Ինդեքս այլևս պետք չէ։

````python
for student_name, grade in class_grades.items():
    if grade >= PASS_MARK:
        print(f"{student_name}: անցավ")
    else:
        print(f"{student_name}: չանցավ")
````

Համեմատի՛ր 14-րդ օրվա տարբերակի հետ.

```python
for position in range(len(student_names)):
    if class_grades[position] >= PASS_MARK:
        print(f"{student_names[position]}: անցավ")
```

Նույն արդյունքը, բայց առանց `position`-ի, առանց `range`-ի, առանց `len`-ի, և առանց
այն վտանգի, որ երկու ցուցակը իրարից անջատվեն։

Չանցածներին հավաքելը՝ նույնպես ավելի պարզ։

````python
failing_students = []

for student_name, grade in class_grades.items():
    if grade < PASS_MARK:
        failing_students.append(student_name)

print("Չեն անցել՝", failing_students)
````

Իսկ միջինը՝ ուղիղ 13-րդ օրվա ձևով։

````python
total = 0

for student_name, grade in class_grades.items():
    total = total + grade

print(f"Դասարանի միջինը՝ {total / len(class_grades):.1f}")
````

## Բ մաս: Մի քանի գնահատական մեկ աշակերտի համար

Իրական մատյանում աշակերտը մեկ գնահատական չունի։ Նա ունի **ցուցակ**։

Եվ բառարանի արժեքը կարող է լինել ցուցակ։

````python
class_grades = {
    "Անի": [9, 10, 8],
    "Դավիթ": [6, 5, 7],
    "Նարե": [10, 10, 9],
    "Արամ": [3, 4, 2],
}

print(class_grades["Անի"])
````

`class_grades["Անի"]` այժմ ցուցակ է։ Ուրեմն նրա հետ կարելի է անել այն ամենը, ինչ
ցուցակի հետ։

````python
print(len(class_grades["Անի"]))
print(class_grades["Անի"][0])
````

Երկու քառակուսի փակագիծ. առաջինը՝ բառարանից աշակերտին, երկրորդը՝ ցուցակից գնահատականը։

Նոր գնահատական ավելացնելը՝ `.append()`, ուղիղ 11-րդ օրվա ձևով։

````python
class_grades["Անի"].append(7)

print(class_grades["Անի"])
````

## Գ մաս: Ամբողջական հաշվետվություն

Հիմա ամեն աշակերտի համար հաշվում ենք իր միջինը։

````python
for student_name, grades in class_grades.items():
    total = 0
    for grade in grades:
        total = total + grade
    student_average = total / len(grades)

    print(f"{student_name}: {grades} — միջինը {student_average:.1f}")
````

Ցիկլ ցիկլի ներսում. արտաքինը՝ աշակերտների վրայով, ներքինը՝ մեկ աշակերտի
գնահատականների վրայով։

Ավելացնենք նաև «անցավ/չանցավ»-ը՝ լրիվ մատյան։

````python
print("=== ՄԱՏՅԱՆ ===")
print()

for student_name, grades in class_grades.items():
    total = 0
    for grade in grades:
        total = total + grade
    student_average = total / len(grades)

    if student_average >= PASS_MARK:
        result = "անցավ"
    else:
        result = "չանցավ"

    print(f"{student_name:<10} {grades}  միջինը՝ {student_average:.1f}  — {result}")

print()
print(f"Ընդամենը՝ {len(class_grades)} աշակերտ")
````

> `{student_name:<10}` — «գրի՛ր անունը և լրացրո՛ւ բացատներով մինչև 10 նիշ»։ Դրա շնորհիվ
> սյունակները հավասարվում են։ Սա միակ ձևավորման հնարքն է, որ պետք կգա։

## Ինչն է դեռ մեզ նյարդայնացնում

Նայի՛ր վերջին երեք բջիջին։ **Այս վեց տողը գրված է երեք անգամ.**

```python
total = 0
for grade in grades:
    total = total + grade
student_average = total / len(grades)
```

Եվ 13-րդ օրը նույնը գրեցինք։ Եվ 14-րդ օրը՝ նույնպես։

Հիմա պատկերացրու, որ սխալ ես գտել այնտեղ — օրինակ՝ պետք է կլորացնել։
**Պետք է ուղղել բոլոր չորս տեղում։** Իսկ եթե մեկը բաց թողնես, երկու տեղում տարբեր
արդյունք կստանաս և չես հասկանա, թե ինչու։

Ծանո՞թ է։ Սա 9-րդ օրվա խնդիրն է՝ `4` թիվը քսանհինգ անգամ։

**Բայց այս անգամ խնդիրը թիվ չէ — խնդիրը վեց տողանոց տրամաբանություն է։**
Փոփոխականը դա չի լուծի։

անցավ/չանցավ։<br/><br/>
<b>5. Ցանկության դեպքում։</b> Հաշվի՛ր, թե քանի աշակերտի միջինն է դասարանի միջինից բարձր։

````python
# Առաջադրանք 1 — քո դասարանը, ամեն աշակերտ երեք գնահատականով

class_grades = {
    "Անի": [9, 10, 8],
    "Դավիթ": [6, 5, 7],
}

print(class_grades)
````

````python
# Առաջադրանք 3 — ամեն աշակերտի միջինը

for student_name, grades in class_grades.items():
    total = 0
    for grade in grades:
        total = total + grade
    print(f"{student_name}: {total / len(grades):.1f}")
````

## Ինչի հասանք

- `for name, grade in class_grades.items():` — ցիկլ բառարանի վրայով, **երկու անուն**։
- Ինդեքսն ու `range`-ը այլևս պետք չեն։
- Բառարանի արժեքը կարող է լինել **ցուցակ**՝ մեկ աշակերտ, մի քանի գնահատական։
- `{name:<10}` — հավասարեցնում է սյունակները։

## Ինչն է դեռ մեզ նյարդայնացնում

Միջին հաշվելու վեց տողը գրված է **չորս տարբեր տեղում**։ Մեկ ուղղումը՝ չորս խմբագրում։
Եվ այս անգամ փոփոխականը չի օգնի — կրկնվողը թիվ չէ, այլ **տրամաբանություն**։
