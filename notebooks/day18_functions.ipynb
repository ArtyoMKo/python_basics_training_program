## Ա մաս: `def` — կոդին անուն տալ

5-րդ օրը **արժեքին** անուն տվեցինք՝ `PASS_MARK = 4`։

Այսօր **կոդին** ենք անուն տալիս։

```python
def say_hello():
    print("Բարև, ուսուցիչ")
```

Այս բջիջը ոչինչ չտպեց։ **Եվ դա ճիշտ է։**

Մենք ոչինչ չենք արել — մենք միայն ասացինք. «այս երկու տողը այսուհետ կոչվում է
`say_hello`»։

Կոդը աշխատեցնելու համար պետք է այն **կանչել**։

```python
say_hello()
```

> **Սահմանելը և կանչելը տարբեր բաներ են։**
>
> `def say_hello():` — սահմանում է։ Ոչինչ չի կատարվում։
> `say_hello()` — կանչում է։ Հիմա կատարվում է։
>
> Սա սկզբում ամենաշատ շփոթեցնողն է։ Եթե ֆունկցիան «չի աշխատում», առաջին հարցը՝
> **կանչե՞լ ես այն**։

Կանչել կարելի է որքան ուզես անգամ։

```python
say_hello()
say_hello()
say_hello()
```

## Բ մաս: Պարամետր

Ֆունկցիան օգտակար է դառնում, երբ ամեն կանչի ժամանակ **ուրիշ արժեքով** է աշխատում։

Արժեքը գրում ենք փակագծերի մեջ։

```python
def greet_student(student_name):
    print(f"Բարև, {student_name}")

greet_student("Անի")
greet_student("Դավիթ")
```

`student_name`-ը **պարամետր** է. անուն, որը գոյություն ունի միայն ֆունկցիայի ներսում։
Ամեն կանչի ժամանակ այն ստանում է այն արժեքը, որ գրել ես փակագծերում։

Դա նույն գաղափարն է, ինչ ցիկլի փոփոխականը 12-րդ օրը։

Պարամետրերը կարող են մի քանիսը լինել։

```python
def print_student_row(student_name, grade):
    print(f"{student_name:<10} {grade}")

print_student_row("Անի", 9)
print_student_row("Դավիթ", 6)
print_student_row("Նարե", 10)
```

**Հերթականությունը կարևոր է։** Առաջին արժեքը գնում է առաջին պարամետրին։

```python
print_student_row(9, "Անի")
```

Ոչ մի սխալ չեղավ — պարզապես անհեթեթ արդյունք։ Python-ը չգիտի, թե ինչ ես նկատի ունեցել։

## Գ մաս: Երեկվա վեց տողը

Ահա այն, վերջին անգամ գրված.

```python
def print_average(grades):
    total = 0
    for grade in grades:
        total = total + grade
    student_average = total / len(grades)
    print(f"Միջինը՝ {student_average:.1f}")
```

Եվ հիմա՝ չորս կանչ։

```python
print_average([9, 10, 8])
print_average([6, 5, 7])
print_average([10, 10, 9])
print_average([3, 4, 2])
```

Երեկ սա քսանչորս տող էր։ Հիմա՝ չորս։

Իսկ բառարանի հետ միասին՝

```python
class_grades = {
    "Անի": [9, 10, 8],
    "Դավիթ": [6, 5, 7],
    "Նարե": [10, 10, 9],
    "Արամ": [3, 4, 2],
}

for student_name, grades in class_grades.items():
    print(student_name)
    print_average(grades)
```

## Մեկ սխալ, մեկ ուղղում

**Ահա այսօրվա իրական դասը։**

Ենթադրենք, տնօրենը ասում է. «միջինը գրե՛ք երկու նիշով, ոչ թե մեկ»։

Երեկ դա չորս խմբագրում կլիներ, և մեկը բաց թողնելը շատ հավանական էր։

Այսօր՝ **մեկ խմբագրում**։ Փոխի՛ր `:.1f`-ը `:.2f`-ի ներքևի բջիջում և գործարկի՛ր
հաջորդ բջիջը՝ առանց այն դիպչելու։

```python
def print_average(grades):
    total = 0
    for grade in grades:
        total = total + grade
    student_average = total / len(grades)
    print(f"Միջինը՝ {student_average:.2f}")
```

```python
for student_name, grades in class_grades.items():
    print(student_name)
    print_average(grades)
```

Բոլոր չորս տեղում փոխվեց։ Որովհետև **տեղը մեկն էր**։

> **Կանոն, զույգ 9-րդ օրվա կանոնին.**
> 9-րդ օր՝ եթե **թիվը** գրված է մեկից ավելի անգամ — տո՛ւր նրան անուն։
> Այսօր՝ եթե **կոդը** գրված է մեկից ավելի անգամ — տո՛ւր նրան անուն։

```python
# Առաջադրանք 1

def greet_student(student_name):
    print(f"Բարև, {student_name}")

greet_student("Անի")
```

```python
# Առաջադրանք 2 — մեկ տող մատյանից, կանչված ցիկլի ներսում

def print_student_row(student_name, grade):
    print(f"{student_name:<10} {grade}")

simple_grades = {"Անի": 9, "Դավիթ": 6, "Նարե": 10}

for student_name, grade in simple_grades.items():
    print_student_row(student_name, grade)
```

## Ինչի հասանք

- `def name(...):` — կոդին անուն է տալիս։ **Ոչինչ չի կատարվում, քանի դեռ չես կանչել։**
- `name(...)` — կանչում է։ Որքան ուզես անգամ։
- Պարամետրերը թույլ են տալիս ամեն կանչին ուրիշ արժեք տալ։ **Հերթականությունը կարևոր է։**
- **Եթե կոդը գրված է մեկից ավելի անգամ — տո՛ւր նրան անուն։**
