## Ինչի հասանք

- Քառասուն փոփոխական՝ գրված ձեռքով։
- Քառասուն փոփոխական՝ ուղղված ձեռքով։
- Աշխատող կոդ։ **Ամեն ինչ ճիշտ է։ Ոչ մի սխալ չկա։**

## Ինչն է դեռ մեզ նյարդայնացնում

Ամեն ինչ աշխատեց։ Ոչ մի կարմիր տեքստ չտեսանք։ Եվ այնուամենայնիվ սա վատ էր.

- **Քառասուն անուն քառասուն արժեքի համար։** Նրանք բոլորը նույն բանն են՝ գնահատական։
- **Մեկ փոփոխություն = քառասուն խմբագրում։** Իսկ եթե դասարանում 300 աշակերտ լիներ,
  կլիներ 300 խմբագրում։
- **Չես կարող ասել «տպիր բոլորը»։** Ամեն մեկը պետք է անվանես առանձին։
- **Մի տեղ սխալվելը հեշտ է, գտնելը՝ ոչ։** Եթե 17-րդ աշակերտին մոռացել ես ավելացնել,
  ոչ մի սխալ չես տեսնի։ Պարզապես թիվը սխալ կլինի։

**Այս ամենի լուծումը գոյություն ունի, և այն պարզ է։** Բայց այսօր չենք սովորելու։

Այսօր կարևորն այն է, որ **զգաս խնդիրը**, որովհետև այդ դեպքում լուծումը հիշելու կարիք
չի ունենա — այն ուղղակի իմաստ կունենա։
