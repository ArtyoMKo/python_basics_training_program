## Հարց 6 — Մեկ ցիկլ, շատ տեսակ

> **Ի՞նչ է չափում՝** բազմաձևություն — ցիկլ **առանց** `if`-ի
> **Միավոր՝** 5
> `isinstance`-ով կամ `if`-ով գրված ճիշտ պատասխանը՝ միավորի կեսը։
> Հարցը ստուգում է 10-րդ թեմայի գաղափարը։

### Տարբերակ A, B և C

**Ցիկլի ներսում `if` չպետք է լինի։**
