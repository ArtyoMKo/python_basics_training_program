## Ա մաս: Բջիջը

Այս տետրը բաղկացած է **բջիջներից**։ Այս մեկը, որ հիմա կարդում ես, տեքստային բջիջ է։

Ներքևինը՝ **կոդի բջիջ** է։ Կտտացրու նրա վրա և սեղմի՛ր **Shift + Enter**։

```python
2 + 2
```

Համակարգիչը հաշվեց և պատասխանը գրեց բջիջի տակ։

Ուշադրություն դարձրու՝ բջիջի ձախ կողմում հայտնվեց `[1]`։ Դա նշանակում է՝ «սա գործարկվել է
առաջինը»։ Երբ ինչ-որ բան չի աշխատում, այս թվերը ցույց են տալիս, թե ինչ հերթականությամբ ես
գործարկել բջիջները։

**Փոխի՛ր վերևի բջիջը** — գրի՛ր `10 * 5` — և նորից սեղմի՛ր Shift + Enter։

```python
10 * 5
```

## Բ մաս: `print` — համակարգիչը խոսում է

Հաշվելը լավ է, բայց մեզ պետք է, որ համակարգիչը **գրի**։ Դրա համար կա `print`։

```python
print("Hello")
```

Երեք բան կատարվեց այս մեկ տողում.

- `print` — հրահանգի անունը։ Միշտ փոքրատառ։
- `( )` — փակագծերի մեջ դնում ենք այն, ինչ պետք է տպել։
- `" "` — չակերտների մեջ դնում ենք **տեքստը**։

Չակերտները պարտադիր են։ Առանց դրանց Python-ը կկարծի, որ `Hello`-ն հրահանգ է, և չի գտնի այն։

> **Ինչո՞ւ է կոդն անգլերեն։** Բացատրությունները հայերեն են և կմնան հայերեն։ Բայց կոդի
> ներսում ամեն բան անգլերեն է՝ բառերը, անունները, տեքստերը։ Այդպես այն կոդը, որ գրում ես
> այստեղ, նույնությամբ նման է աշխարհի ցանկացած Python-ի, և կարող ես ինտերնետում պատասխան
> փնտրել, երբ խնդիր ունենաս։

```python
print("7-B class register")
```

`print`-ը կարող է տպել մի քանի բան միանգամից, եթե դրանք բաժանես ստորակետով։
Ուշադրություն դարձրու՝ բացատը Python-ն ինքն է դնում։

```python
print("Ani", 9)
```

Իսկ դատարկ `print()`-ը պարզապես դատարկ տող է թողնում։ Դա պետք կգա, երբ մատյան ենք տպելու։

```python
print("Student: Ani")
print()
print("Grade: 9")
```

## Ինչի հասանք

- Համակարգչի վրա տեղադրված է այն ամենը, ինչ պետք է ամբողջ դասընթացի համար։ **Այլևս ոչինչ չենք տեղադրելու։**
- Տետրը բաղկացած է բջիջներից, բջիջը գործարկվում է **Shift + Enter**-ով։

- Բացատրությունները հայերեն են, կոդը՝ անգլերեն։
