## Ա մաս: Չորս տեսակը

Դասարանի մատյանում կան ուղիղ չորս տեսակի տվյալ։

| Ինչ | Օրինակ | Անգլերեն անունը |
|---|---|---|
| Անուն | `"Ani"` | `str` — տեքստ |
| Գնահատական | `9` | `int` — ամբողջ թիվ |
| Միջին | `6.5` | `float` — տասնորդական թիվ |
| Անցա՞վ | `True` | `bool` — ճիշտ կամ սխալ |

````python
print("Ani")
print(9)
print(6.5)
````

Ուշադրություն՝ տասնորդական թիվը գրվում է **կետով**, ոչ թե ստորակետով։ `6.5`, ոչ թե `6,5`։

Չորրորդ տեսակը ամենափոքրն է. այն ունի ընդամենը **երկու հնարավոր արժեք**։

````python
print(True)
print(False)
````

`True` և `False` — մեծատառով, առանց չակերտների։ Սրանք Python-ի բառերն են, ոչ թե տեքստ։

Այս երկուսը հիմա անօգուտ են թվում։ 9-րդ թեմայում դրանք կդառնան ամենակարևորը՝ «անցա՞վ, թե՞ ոչ»։

## Բ մաս: `type()` — հարցնել, թե ինչ է սա

````python
print(type("Ani"))
print(type(9))
print(type(6.5))
print(type(True))
````

Այս չորս բառը — `str`, `int`, `float`, `bool` — կհանդիպեն սխալի հաղորդագրություններում։
Երբ կարդում ես կարմիր տեքստը և տեսնում `str`, հիմա գիտես, որ խոսքը տեքստի մասին է։

## Գ մաս: Ի՞նչ է լինում, երբ խառնում ես

````python
print(9 + 1)
print(9 * 2)
print(78 / 12)
````

Ուշադրություն վերջին տողին՝ `78 / 12` տվեց `6.5`։ **Բաժանումը միշտ տալիս է
տասնորդական թիվ**, նույնիսկ երբ արդյունքը կլոր է։

````python
print(10 / 2)
print(type(10 / 2))
````

`5.0`, ոչ թե `5`։ Սա սովորական է և խանգարել չի տա — պարզապես իմացի՛ր, որ այդպես է։

Իսկ տեքստի հետ ի՞նչ է լինում, երբ գումարում ես։

````python
print("Ani" + "Davit")
````

Տեքստերի դեպքում `+`-ը **կպցնում է** դրանք իրար։ Բացատ չի դնում — եթե պետք է, ինքդ ավելացրու։

````python
print("Ani" + " and " + "Davit")
````

### Հիմա կոտրենք

Իսկ ի՞նչ է լինում, եթե տեքստ ու թիվ գումարենք։

**Նախ գուշակի՛ր։** Ի՞նչ ես կարծում՝ `"9" + 1` ինչ կտա։ Գրի՛ր պատասխանդ թղթի վրա,
հետո գործարկի՛ր բջիջը։

````python
print("9" + 1)
````

### Ինչ գրված է այնտեղ

```
TypeError: can only concatenate str (not "int") to str
```

| Մասը | Ի՞նչ է ասում |
|---|---|
| `TypeError` | «Այս երկուսը միասին չեն աշխատում» |
| `str` | տեքստ |
| `int` | ամբողջ թիվ |
| ամբողջը | «Տեքստին կարող եմ միայն տեքստ կպցնել, ոչ թե թիվ» |

Python-ը չի ենթադրում, թե ինչ ես ուզում։ `"9" + 1` — արդյունքը պետք է լինի `10`,
թե՞ `"91"`։ Երկուսն էլ տրամաբանական են, ուստի նա հրաժարվում է գուշակել։

````python
print("9" + "1")
print(9 + 1)
````

**Ահա թե ինչու է տեսակը կարևոր։** Նույն `+` նշանը երկու բոլորովին տարբեր բան է անում՝
կախված նրանից, թե ինչ տեսակի արժեքներ ես տվել։

## Ինչի հասանք

- Չորս տեսակ, և ամբողջ դասընթացում ուրիշ բան չկա՝ `str`, `int`, `float`, `bool`։
- `type(...)` — հարցնում է, թե ինչ տեսակի է արժեքը։
- Բաժանումը միշտ տալիս է տասնորդական թիվ։
- `+`-ը թվերի համար գումարում է, տեքստերի համար՝ կպցնում։ Խառնելը տալիս է `TypeError`։

## Ի՞նչ է գալիս հետո

Այսօր տեսանք, որ `"9"`-ը և `9`-ը տարբեր բաներ են։ Ընդմիջումից հետո կսովորենք մեկը մյուսին
վերածել — և կտեսնենք, թե ինչու է դա **անհրաժեշտ**, երբ ուզում ենք ինչ-որ բան հարցնել։
