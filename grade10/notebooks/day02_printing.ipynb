## Ա մաս: Մի քանի արժեք մեկ տողում

Այս նիստի սկզբում տեսանք, որ `print`-ը կարող է մի քանի բան տպել, եթե բաժանես ստորակետով։
Հիմա օգտագործենք դա մատյանի տողերի համար։

````python
print("Ani", 9)
````

Python-ը ինքն է բացատ դնում արժեքների միջև։ Բայց մատյանում մեզ պետք է երկու կետ։
Այն էլ պարզապես ավելացնում ենք որպես տեքստ։

````python
print("Ani:", 9)
````

Նկատի՛ր տարբերությունը. `"Ani:"` — չակերտների մեջ է, ուրեմն տեքստ է։
`9` — չակերտների մեջ չէ, ուրեմն թիվ է։ Հաջորդ նիստում կտեսնենք, թե ինչու է դա կարևոր։

````python
print("Student:", "Ani", "| Grade:", 9)
````

## Բ մաս: Մեկնաբանություններ

`#` նշանից հետո գրվածը Python-ը **չի կարդում**։ Դա գրառում է քեզ համար։

Սա շատ ավելի օգտակար է, քան թվում է. երեք շաբաթ հետո չես հիշի, թե ինչու ես այդպես գրել։

````python
# The register header. Change it to your own class.
print("7-B class - biology")
````

Մեկնաբանությունը կարող է լինել նաև տողի վերջում։

Իսկ եթե `#` դնես կոդի առջև, այդ տողը պարզապես չի աշխատի։ Սա շատ հարմար է, երբ ուզում ես
ժամանակավորապես անջատել մի տող՝ առանց ջնջելու։

````python
print("This line runs")
# print("This one does not")
````

## Գ մաս: Կոտրենք դիտավորյալ

Հիմա ամենակարևոր մասը։ **Կարմիր տեքստը սխալ չէ քո կողմից — այն հաղորդագրություն է։**

Ներքևի բջիջում **դիտավորյալ** մոռացել ենք փակել չակերտը։ Գործարկի՛ր այն և նայի՛ր, թե
ինչ է գրում Python-ը։

````python
print("Ani: 9)
````

### Ինչ գրված է այնտեղ

Կարմիր տեքստը երկար է, բայց **կարդալու է միայն վերջին տողը**։

```
SyntaxError: unterminated string literal
```

| Մասը | Ի՞նչ է ասում |
|---|---|
| `SyntaxError` | «Չեմ հասկանում այս տողը» |
| `unterminated string` | «Չակերտը բացվեց, բայց չփակվեց» |
| Սլաքը `^` | Ցույց է տալիս, թե որտեղից սկսած չի հասկանում |

Python-ը ոչ միայն ասում է, որ սխալ կա, այլև՝ **որ տողում** և **ինչ տեսակի**։

> **Սխալի հաղորդագրությունները անգլերեն են և միշտ կլինեն։** Ոչ մի ծրագրավորման լեզու
> դրանք հայերեն չի տալիս։ Ուստի դասընթացի ընթացքում սովորում ենք նաև դրանք կարդալ —
> բոլորի ցանկը կա հուշաթերթի վերջում։

Ուղղենք՝ ավելացնում ենք բաց մնացած չակերտը։

````python
print("Ani: 9")
````

## Ինչի հասանք

- `print`-ը կարող է տպել մի քանի արժեք՝ ստորակետով բաժանված։
- Չակերտների մեջ գրվածը տեքստ է. առանց չակերտների գրված թիվը՝ թիվ։
- `#`-ից հետո գրվածը Python-ը չի կարդում — դա գրառում է քեզ համար։
- **Կարմիր տեքստը հաղորդագրություն է։** Կարդում ենք վերջին տողը. այնտեղ է սխալի անունը։

## Ի՞նչ է գալիս հետո

Այսօր երկու անգամ ասացինք՝ «սա տեքստ է, իսկ սա՝ թիվ», բայց չբացատրեցինք, թե ինչու է դա
կարևոր։ Հաջորդ նիստում կտեսնենք, թե ինչ է լինում, երբ դրանք իրար խառնում ես։
