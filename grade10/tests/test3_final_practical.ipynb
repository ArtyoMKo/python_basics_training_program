## Հարց 8 — Նոր խնդիր

> **Ուսուցչի նշում — մի՛ ցուցադրեք մասնակցին հարթակում.**
> **Հարցի տեսակ՝** փոխանցման հարց (transfer)
> **Ի՞նչ է չափում՝** արդյոք կոդ գրելու վարժությունը ազատե՞լ է ուշադրությունը խնդիր լուծելու համար
> **Միավոր՝** 10 — **գնահատվում է առանձին և չի գումարվում 70-ին**
>
> **Սա ամբողջ ծրագրի գլխավոր չափումն է։** Խնդիրը դասին ցույց չի տրվել, բայց լուծելու
> համար նոր շարահյուսություն պետք չէ։ Մանրամասն գնահատման կարգը՝ `mark3_guide.md`։
>
> **Արդյունքը չի զեկուցվում որպես «անցածների տոկոս»։** Արձանագրվում է չորս բան՝
> պլանը գրե՞ց, պլանը ճի՞շտ էր, կոդն աշխատե՞ց, որտե՞ղ կանգ առավ։
> **Ճիշտ պլան + անավարտ կոդ = հաջողություն։**

**Հարթակի կանոն.** Բոլոր մասնակիցներին տրվում է **նույն** հարցը՝ առանց A/B/C
տարբերակների։ Այս հարցը տրվում է **վերջում**, մնացած հարցերից հետո։
